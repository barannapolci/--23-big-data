```python
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go

# Завантажуємо дані для двох акцій
ticker_1 = 'TSLA'  # Tesla
ticker_2 = 'MSFT'  # Microsoft

# Завантажуємо ринкові дані для S&P 500 (для розрахунку бета)
market_data = yf.download('^GSPC', start='2023-04-01', end='2024-04-01')

# Завантажуємо ціни для двох акцій
stock_data_1 = yf.download(ticker_1, start='2023-04-01', end='2024-04-01')
stock_data_2 = yf.download(ticker_2, start='2023-04-01', end='2024-04-01')
```

```python
# Обчислюємо денні дохідності
stock_data_1['Daily_Return'] = stock_data_1['Close'].pct_change()
stock_data_2['Daily_Return'] = stock_data_2['Close'].pct_change()

# Обчислюємо кумулятивні дохідності
stock_data_1['Cumulative_Return'] = (1 + stock_data_1['Daily_Return']).cumprod() - 1
stock_data_2['Cumulative_Return'] = (1 + stock_data_2['Daily_Return']).cumprod() - 1
```

```python
# Візуалізація щоденних дохідностей
fig = go.Figure()

fig.add_trace(go.Scatter(x=stock_data_1.index, y=stock_data_1['Daily_Return'],
                         mode='lines', name=ticker_1, line=dict(color='blue')))
fig.add_trace(go.Scatter(x=stock_data_2.index, y=stock_data_2['Daily_Return'],
                         mode='lines', name=ticker_2, line=dict(color='green')))

fig.update_layout(title='Щоденна дохідність акцій',
                  xaxis_title='Дата', yaxis_title='Денна дохідність',
                  legend=dict(x=0.02, y=0.95))

fig.show()
```

```python
# Візуалізація кумулятивних дохідностей
fig = go.Figure()

fig.add_trace(go.Scatter(x=stock_data_1.index, y=stock_data_1['Cumulative_Return'],
                         mode='lines', name=ticker_1, line=dict(color='blue')))
fig.add_trace(go.Scatter(x=stock_data_2.index, y=stock_data_2['Cumulative_Return'],
                         mode='lines', name=ticker_2, line=dict(color='green')))

fig.update_layout(title='Кумулятивна дохідність акцій',
                  xaxis_title='Дата', yaxis_title='Кумулятивна дохідність',
                  legend=dict(x=0.02, y=0.95))

fig.show()
```

```python
# Волатильність для кожної акції (стандартне відхилення)
volatility_1 = stock_data_1['Daily_Return'].std()
volatility_2 = stock_data_2['Daily_Return'].std()

# Виведемо волатильність для порівняння
print(f'Волатильність для {ticker_1}: {volatility_1}')
print(f'Волатильність для {ticker_2}: {volatility_2}')
```

```python
# Обчислюємо денні дохідності для S&P 500
market_data['Daily_Return'] = market_data['Close'].pct_change()

# Обчислюємо бета для кожної акції
cov_1 = stock_data_1['Daily_Return'].cov(market_data['Daily_Return'])
cov_2 = stock_data_2['Daily_Return'].cov(market_data['Daily_Return'])
var_market = market_data['Daily_Return'].var()

beta_1 = cov_1 / var_market
beta_2 = cov_2 / var_market

# Виводимо коефіцієнти бета
print(f'Бета для {ticker_1}: {beta_1}')
print(f'Бета для {ticker_2}: {beta_2}')

# Висновки
if beta_1 > beta_2:
    conclusion = f"Акція {ticker_1} має вищу волатильність (вищий коефіцієнт бета), ніж акція {ticker_2}."
else:
    conclusion = f"Акція {ticker_2} має вищу волатильність (вищий коефіцієнт бета), ніж акція {ticker_1}."

print(conclusion)
```
