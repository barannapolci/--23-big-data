```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium
from folium.plugins import MarkerCluster

# Завантаження даних
df = pd.read_excel("bda_dataset1_3.xlsx")
df.head()
```

```python
map_center = [df['lat'].mean(), df['long'].mean()]
m = folium.Map(location=map_center, zoom_start=10)
marker_cluster = MarkerCluster().add_to(m)

for _, row in df.iterrows():
    folium.Marker(
        location=[row['lat'], row['long']],
        popup=f"Ціна: ${row['price']}<br>Кімнат: {row['bedrooms']}<br>Рік: {row['yr_built']}"
    ).add_to(marker_cluster)

m
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='yr_built', order=df['yr_built'].value_counts().index)
plt.title('Кількість obj за роком побудови')
plt.xlabel('Рік побудови')
plt.ylabel('Кількість obj')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='bedrooms', order=sorted(df['bedrooms'].unique()))
plt.title('Кількість obj за кількістю кімнат')
plt.xlabel('rooms')
plt.ylabel('Кількість obj')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='grade', order=sorted(df['grade'].unique()))
plt.title('Кількість obj за якістю')
plt.xlabel('Оцінка якості')
plt.ylabel('Кількість obj')
plt.show()
```
