```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

data = pd.read_csv('bda_dataset3_2.csv')

# Перевіряємо перші кілька рядків та наявність пропущених значень
print(data.head())
print(data.isnull().sum())
print(data.describe())
```

```python
basket = data.groupby('Member_number')['itemDescription'].apply(list).reset_index()

print(basket.head())
```

```python
basket_encoded = basket['itemDescription'].str.join('|').str.get_dummies('|')

# Перевіряємо перші кілька рядків отриманої бінарної матриці
print(basket_encoded.head())
```

```python

frequent_itemsets = apriori(basket_encoded, min_support=0.02, use_colnames=True)

print(frequent_itemsets.head())
```

```python

# Генеруємо асоціативні правила з показником lift не менше 1
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1)

print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())
```

```python
# Відображаємо 10 правил з найвищим показником lift
top_10_rules = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].sort_values(by='lift', ascending=False).head(10)
print(top_10_rules)
```
