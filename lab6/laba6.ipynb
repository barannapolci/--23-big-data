```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_excel('bda_dataset6_2.xlsx')

# Видаляємо рядки з пропущеними значеннями
df_clean = df.dropna()
```

```python
control = df_clean[df_clean['Campaign Name'] == 'Control']
experiment = df_clean[df_clean['Campaign Name'] == 'Experiment']
```

```python
metrics = {
    'Кількість кліків': '# of Website Clicks',
    'Пошукові запити': '# of Searches',
    'Витрати (USD)': 'Spend [USD]',
    'Додавання до кошика': '# of Add to Cart'
}

results = []

for name, col in metrics.items():
    t_stat, p_val = ttest_ind(experiment[col], control[col], equal_var=False)
    results.append({
        'Показник': name,
        'T-статистика': round(t_stat, 3),
        'P-значення': round(p_val, 4),
        'Є значуща різниця (p < 0.05)': 'Так' if p_val < 0.05 else 'Ні'
    })

comparison = pd.DataFrame(results)
print(comparison)
```

```python
if any(p < 0.05 for p in comparison['P-значення']):
    print("\n Експериментальна маркетингова кампанія була ефективною: хоча б один показник змінився статистично значущо.")
else:
    print("\n Експериментальна кампанія не показала статистично значущих змін порівняно з контрольною.")
```
